```python
for file in files:
    if file.name.endswith(".csv"):
        path = file.path

        df = (
            spark.read
                .format("csv")
                .option("header", "true")
                .option("inferSchema", "true")
                .load(path)
        )

        print("=" * 100)
        print(f"FILE: {file.name}")
        print(f"ROWS: {df.count():,}")
        print(f"COLUMNS: {len(df.columns)}")
        print("COLUMN NAMES:")
        print(df.columns)
```
